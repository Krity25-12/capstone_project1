```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('retail_customer_purchasing_sales_dataset.csv')

print("Dataset loaded successfully!")

df.head()
```

```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Dataset shape:")
print(df.shape)

print("Column names:")
print(df.columns)

print("Dataset information:")
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())
```

```python
# Remove duplicate records
df = df.drop_duplicates()

# Convert Order_Date into datetime format
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

# Check missing values
print(df.isnull().sum())

print("\nData cleaning completed!")
```

```python
print("Statistical Summary:")

display(df.describe())

print("\nTotal Sales:")
print(df['Order_Value'].sum())

print("\nTotal Quantity Sold:")
print(df['Quantity'].sum())

print("\nAverage Order Value:")
print(df['Order_Value'].mean())
```

```python
product_sales = df.groupby('Product_Name')['Quantity'].sum()

product_sales = product_sales.sort_values(
    ascending=False
)

print("Frequently Purchased Products:")

display(product_sales)
```

```python
plt.figure(figsize=(10, 5))

product_sales.plot(kind='bar')

plt.title('Frequently Purchased Products')
plt.xlabel('Product Name')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Category')['Order_Value'].sum()

category_sales = category_sales.sort_values(
    ascending=False
)

print("Sales by Product Category:")

display(category_sales)
```

```python
plt.figure(figsize=(9, 5))

category_sales.plot(
    kind='bar',
    color='skyblue'
)

plt.title('Sales Performance by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
df['Month'] = df['Order_Date'].dt.month

monthly_sales = df.groupby('Month')['Order_Value'].sum()

print("Monthly Sales:")

display(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

monthly_sales.plot(
    kind='line',
    marker='o'
)

plt.title('Monthly Retail Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.grid(True)
plt.xticks(range(1, 13))

plt.show()
```

```python
customer_sales = df.groupby(
    'Customer_Segment'
)['Order_Value'].sum()

print("Sales by Customer Segment:")

display(customer_sales)
```

```python
plt.figure(figsize=(7, 5))

customer_sales.plot(
    kind='bar',
    color='orange'
)

plt.title('Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales')

plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df.groupby('Region')['Order_Value'].sum()

print("Regional Sales:")

display(region_sales)
```

```python
plt.figure(figsize=(8, 5))

region_sales.plot(
    kind='bar',
    color='purple'
)

plt.title('Sales Performance Across Regions')
plt.xlabel('Region')
plt.ylabel('Total Sales')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Quantity',
    y='Order_Value',
    hue='Category'
)

plt.title('Quantity Purchased vs Order Value')
plt.xlabel('Quantity Purchased')
plt.ylabel('Order Value')

plt.show()
```

```python
payment_sales = df.groupby(
    'Payment_Method'
)['Order_Value'].sum()

print("Sales by Payment Method:")

display(payment_sales)
```

```python
plt.figure(figsize=(8, 5))

payment_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution by Payment Method')
plt.ylabel('')

plt.show()
```

```python
correlation = df[
    ['Quantity', 'Unit_Price',
     'Discount_Percent', 'Order_Value']
].corr()

print("Correlation Matrix:")

display(correlation)
```

```python
print("========== RETAIL SALES ANALYSIS ==========")

print("Total Transactions:", len(df))

print("Total Customers:", df['Customer_ID'].nunique())

print("Total Products:", df['Product_Name'].nunique())

print("Total Sales: Rs.", round(df['Order_Value'].sum(), 2))

print("Total Quantity Sold:", df['Quantity'].sum())

print("Average Order Value: Rs.",
      round(df['Order_Value'].mean(), 2))

print("Best Selling Product:",
      product_sales.idxmax())

print("Best Performing Category:",
      category_sales.idxmax())

print("Highest Sales Region:",
      region_sales.idxmax())

print("===========================================")
```
